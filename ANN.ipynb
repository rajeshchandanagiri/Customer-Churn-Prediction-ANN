```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\User\OneDrive\Desktop\Churn_Modelling.csv")
df.head()
```

```python
x = df.iloc[: , 3:13]
y = df.iloc[: ,13]
```

```python
x.head()
```

```python
y
```

```python
geography = pd.get_dummies(x['Geography'], dtype = int , drop_first = True)
geography
```

```python
gender  = pd.get_dummies(x['Gender'] ,dtype = int, drop_first = True )
gender
```

```python
x = x.drop(['Geography','Gender'], axis = 1)
```

```python
x.head()
```

```python
x1 = pd.concat([x, geography , gender], axis = 1)
x1
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test = train_test_split(x1 , y , test_size = 0.2 , random_state = 0)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)

```

```python
x_train.shape
```

```python
x_test
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import LeakyReLU , PReLU , ELU, ReLU
from tensorflow.keras.layers import Dropout
```

```python
classifier = Sequential()
```

```python
# adding the input layers
```

```python
classifier.add(Dense(units = 11, activation = 'relu'))

```

```python
# adding the  first hidden layer
```

```python
classifier.add(Dense(units = 7 , activation = 'relu'))
classifier.add(Dropout(0.2))
```

```python
## adding the second hidden layer
```

```python
classifier.add(Dense(units = 6 , activation = 'relu'))
classifier.add(Dropout(0.3))
```

```python
## adding the output layer
```

```python
classifier.add(Dense(1 , activation = 'sigmoid'))
```

```python
classifier.compile(optimizer = 'adam' , loss = 'binary_crossentropy' , metrics = ['accuracy'])
```

```python
# adam inbuilt provide the learnig rate if we want to add new learnig rate
```

```python
import tensorflow
opt = tensorflow.keras.optimizers.Adam(learning_rate = 0.01)
```

```python
## early stopping  keras 
```

```python
import tensorflow as tf
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    min_delta=0.0001,
    patience=20,
    verbose=1,
    mode="auto",
    baseline=None,
    restore_best_weights=False,
    start_from_epoch=0,
)
```

```python
# we need to train our neural network
```

```python
model_history = classifier.fit(x_train , y_train , validation_split = 0.33 , batch_size =10 , epochs = 1000 , callbacks = early_stopping )
```

```python
model_history.history.keys()
```

```python
y_pred = classifier.predict(x_test)
y_pred = (y_pred >=0.5)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test , y_pred)
cm
```

```python
## calucate the accuracy
```

```python
from sklearn.metrics import accuracy_score
score = accuracy_score(y_pred , y_test)
```
